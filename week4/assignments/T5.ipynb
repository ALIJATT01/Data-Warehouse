## T5 — Cross-filter direction (12 marks)

Six situations. The verdict is **Single**, **Both**, or **solve it at measure level**.
Only ONE of the six is a genuine case for changing anything — and even that one is solved in a
measure, not the model. Zero Both verdicts: Both is not forbidden, it is just never the answer here.

| # | Situation | Verdict | Why |
|---|---|---|---|
| 1 | A slicer on DimMode must filter the shipment totals on every visual | **Single** | Filter flows dimension → fact through the single active relationship |
| 2 | DimCity and DimMode both set to Both; Power BI refuses the second | **Single** | Both on two dimensions creates ambiguity — set both back to Single |
| 3 | A slicer on DimCity should grey out modes never used from that city | **Measure-level** | Keep the model Single; visual-level "show items with no data" does it |
| 4 | A card showing how many cities appear in FactShipment at all | **Measure-level** | `DISTINCTCOUNT(FactShipment[OriginCityKey])` — a measure, not a model change |
| 5 | RLS restricts users to their own province | **Single** | RLS filters flow dimension → fact; Both would leak fact filters back |
| 6 | A second fact, FactFuelSpend, shares DimCity and DimDate | **Single** | Two facts sharing dimensions must both stay Single — Both = ambiguous paths |
